```python

from pathlib import Path
import yaml
import cv2
import numpy as np
import shutil

# Locate dataset configuration
search_root = Path("/content/drowsiness_dataset")
yaml_files = list(search_root.rglob("data.yaml"))

if not yaml_files:
    raise FileNotFoundError(
        "Dataset not found. Upload and extract your ZIP first."
    )

yaml_path = yaml_files[0]
ROOT = yaml_path.parent

with open(yaml_path, "r") as f:
    config = yaml.safe_load(f)

names = config["names"]
if isinstance(names, dict):
    names = [names[i] for i in sorted(names)]
names = [str(n) for n in names]

print("Dataset:", ROOT)
print("Classes:", names)

OUT = Path("/content/vgg19_crops")
if OUT.exists():
    shutil.rmtree(OUT)

extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
splits = {"train": "train", "valid": "valid", "test": "test"}

for split, folder_name in splits.items():
    image_dir = ROOT / folder_name / "images"
    label_dir = ROOT / folder_name / "labels"

    if not image_dir.exists() or not label_dir.exists():
        raise FileNotFoundError(
            f"Missing images or labels for split: {split}"
        )

    image_lookup = {
        p.stem: p for p in image_dir.rglob("*")
        if p.is_file() and p.suffix.lower() in extensions
    }

    counts = {name: 0 for name in names}

    for label_file in label_dir.rglob("*.txt"):
        image_path = image_lookup.get(label_file.stem)
        if image_path is None:
            continue

        image = cv2.imread(str(image_path))
        if image is None:
            continue

        h, w = image.shape[:2]

        for idx, line in enumerate(label_file.read_text().splitlines()):
            parts = line.split()
            if len(parts) < 5:
                continue

            try:
                class_id = int(float(parts[0]))
                xc, yc, bw, bh = map(float, parts[1:5])
            except ValueError:
                continue

            if not 0 <= class_id < len(names):
                continue
            if not np.isfinite([xc, yc, bw, bh]).all():
                continue

            x1 = max(0, int((xc - bw / 2) * w))
            y1 = max(0, int((yc - bh / 2) * h))
            x2 = min(w, int((xc + bw / 2) * w))
            y2 = min(h, int((yc + bh / 2) * h))

            if x2 <= x1 or y2 <= y1:
                continue

            crop = image[y1:y2, x1:x2]
            class_name = names[class_id]

            save_dir = OUT / split / class_name
            save_dir.mkdir(parents=True, exist_ok=True)

            save_path = save_dir / f"{label_file.stem}_{idx}.jpg"
            if cv2.imwrite(str(save_path), crop):
                counts[class_name] += 1

    print(f"{split} crops:", counts)

print("\nCrops saved to:", OUT)
```

```python

from tensorflow.keras.applications import VGG19
from tensorflow.keras.layers import (
    Input, GlobalAveragePooling2D, Dense, Dropout
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

num_classes = train_gen.num_classes

base_model = VGG19(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

inputs = Input(shape=(224, 224, 3))
x = base_model(inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation="relu")(x)
x = Dropout(0.4)(x)
outputs = Dense(num_classes, activation="softmax")(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

from tensorflow.keras.callbacks import (
    ModelCheckpoint, ReduceLROnPlateau, EarlyStopping
)

callbacks = [
    ModelCheckpoint(
        "/content/best_vgg19.keras",
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=3,
        min_lr=1e-7,
        verbose=1
    ),
    EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True,
        verbose=1
    )
]

history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=20,
    callbacks=callbacks
)
```

```python

from pathlib import Path

DATASET_PATH = Path("/content/resnet50_crops")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        "Cropped dataset not found. Recreate the crops "
        "from your YOLO dataset first."
    )

IMG_SIZE = (260, 260)
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
    rotation_range=12,
    width_shift_range=0.08,
    height_shift_range=0.08,
    zoom_range=0.12,
    horizontal_flip=True,
    fill_mode="nearest"
)

test_datagen = ImageDataGenerator()

train_gen = train_datagen.flow_from_directory(
    str(DATASET_PATH / "train"),
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True,
    seed=42
)

val_gen = test_datagen.flow_from_directory(
    str(DATASET_PATH / "valid"),
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

test_gen = test_datagen.flow_from_directory(
    str(DATASET_PATH / "test"),
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

assert train_gen.class_indices == val_gen.class_indices == test_gen.class_indices

print("Classes:", train_gen.class_indices)
print("Training crops:", train_gen.samples)
print("Validation crops:", val_gen.samples)
print("Testing crops:", test_gen.samples)
```

```python

base_model = EfficientNetB2(
    weights="imagenet",
    include_top=False,
    input_shape=(260, 260, 3)
)

base_model.trainable = False

inputs = Input(shape=(260, 260, 3))

x = base_model(inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation="relu")(x)
x = BatchNormalization()(x)
x = Dropout(0.4)(x)

outputs = Dense(
    train_gen.num_classes,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

classes = np.unique(train_gen.classes)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_gen.classes
)

class_weights = {
    int(c): float(w)
    for c, w in zip(classes, weights)
}

callbacks = [
    ModelCheckpoint(
        "/content/efficientnetb2_best.keras",
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=3,
        min_lr=1e-7,
        verbose=1
    ),
    EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True,
        verbose=1
    )
]

history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=25,
    class_weight=class_weights,
    callbacks=callbacks
)
```

```python

!pip install -q tensorflow scikit-learn seaborn

import os
import gc
import numpy as np
import tensorflow as tf

from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import (
    ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
)
from tensorflow.keras.applications import VGG19, ResNet50, EfficientNetB2
from tensorflow.keras.applications.vgg19 import preprocess_input as vgg_preprocess
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_preprocess

tf.keras.utils.set_random_seed(42)

BATCH_SIZE = 16
EPOCHS_HEAD = 12
EPOCHS_FINE = 10
OUTPUT_DIR = "/content/drowsiness_ensemble"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# EfficientNetB2 includes its own input rescaling in current Keras.
CONFIGS = {
    "VGG19": {
        "builder": VGG19,
        "size": (224, 224),
        "preprocess": vgg_preprocess,
        "fine_prefix": "block5_",
        "filename": "vgg19_best.keras"
    },
    "ResNet50": {
        "builder": ResNet50,
        "size": (224, 224),
        "preprocess": resnet_preprocess,
        "fine_prefix": "conv5_",
        "filename": "resnet50_best.keras"
    },
    "EfficientNetB2": {
        "builder": EfficientNetB2,
        "size": (260, 260),
        "preprocess": None,
        "fine_prefix": None,
        "filename": "efficientnetb2_best.keras"
    }
}

def make_generators(size, preprocess):
    train_aug = ImageDataGenerator(
        preprocessing_function=preprocess,
        rotation_range=10,
        width_shift_range=0.08,
        height_shift_range=0.08,
        zoom_range=0.10,
        horizontal_flip=True
    )
    eval_aug = ImageDataGenerator(preprocessing_function=preprocess)

    train = train_aug.flow_from_directory(
        str(DATASET_PATH / "train"),
        target_size=size,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=True,
        seed=42
    )
    val = eval_aug.flow_from_directory(
        str(DATASET_PATH / "valid"),
        target_size=size,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=False
    )
    test = eval_aug.flow_from_directory(
        str(DATASET_PATH / "test"),
        target_size=size,
        batch_size=BATCH_SIZE,
        class_mode="categorical",
        shuffle=False
    )
    return train, val, test

def build_model(name, cfg, num_classes):
    size = cfg["size"]
    base = cfg["builder"](
        weights="imagenet",
        include_top=False,
        input_shape=(*size, 3)
    )
    base.trainable = False

    inputs = layers.Input(shape=(*size, 3))
    x = base(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = Model(inputs, outputs, name=name)
    model.compile(
        optimizer=Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

model_paths = {}

for name, cfg in CONFIGS.items():
    print("\n" + "=" * 55)
    print("Training:", name)

    tf.keras.backend.clear_session()
    gc.collect()

    train_gen, val_gen, test_gen = make_generators(
        cfg["size"], cfg["preprocess"]
    )

    assert train_gen.class_indices == val_gen.class_indices
    assert train_gen.class_indices == test_gen.class_indices

    model = build_model(name, cfg, train_gen.num_classes)
    save_path = os.path.join(OUTPUT_DIR, cfg["filename"])

    callbacks = [
        ModelCheckpoint(
            save_path, monitor="val_loss",
            save_best_only=True, verbose=1
        ),
        ReduceLROnPlateau(
            monitor="val_loss", factor=0.2,
            patience=2, min_lr=1e-7, verbose=1
        ),
        EarlyStopping(
            monitor="val_loss", patience=4,
            restore_best_weights=True, verbose=1
        )
    ]

    model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=EPOCHS_HEAD,
        callbacks=callbacks,
        verbose=1
    )

    # Load the best head-trained checkpoint.
    model = tf.keras.models.load_model(save_path)
    base = model.layers[1]
    base.trainable = True

    # Fine-tune deeper layers; keep BatchNorm frozen.
    prefix = cfg["fine_prefix"]
    for layer in base.layers:
        layer.trainable = (
            prefix is not None
            and layer.name.startswith(prefix)
            and not isinstance(layer, layers.BatchNormalization)
        )

    model.compile(
        optimizer=Adam(learning_rate=1e-5),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=EPOCHS_FINE,
        callbacks=callbacks,
        verbose=1
    )

    model_paths[name] = save_path
    print("Best checkpoint saved:", save_path)

    del model, train_gen, val_gen, test_gen
    gc.collect()

print("\nAll three models trained.")
print(model_paths)
```

```python

import pandas as pd
import tensorflow as tf
import numpy as np

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)
import matplotlib.pyplot as plt
import seaborn as sns

probabilities = {}
y_true = None
class_names = None

for name, cfg in CONFIGS.items():
    print("Evaluating:", name)

    _, _, test_gen = make_generators(
        cfg["size"], cfg["preprocess"]
    )

    if class_names is None:
        class_names = [
            n for n, i in sorted(
                test_gen.class_indices.items(),
                key=lambda item: item[1]
            )
        ]
        y_true = test_gen.classes.copy()
    else:
        assert test_gen.class_indices == {
            n: i for i, n in enumerate(class_names)
        }, "Class order differs between generators."
        assert np.array_equal(y_true, test_gen.classes)

    model = tf.keras.models.load_model(model_paths[name])
    test_gen.reset()
    probabilities[name] = model.predict(test_gen, verbose=1)

    del model, test_gen
    tf.keras.backend.clear_session()

# Equal-weight soft voting
ensemble_probs = np.mean(
    [probabilities[n] for n in CONFIGS.keys()],
    axis=0
)

predictions = {
    name: np.argmax(probabilities[name], axis=1)
    for name in CONFIGS
}
predictions["Ensemble"] = np.argmax(ensemble_probs, axis=1)

rows = []
for name, pred in predictions.items():
    rows.append({
        "Model": name,
        "Accuracy (%)": round(100 * accuracy_score(y_true, pred), 2),
        "Precision (%)": round(
            100 * precision_score(y_true, pred, pos_label=1,
                                  zero_division=0), 2
        ),
        "Recall (%)": round(
            100 * recall_score(y_true, pred, pos_label=1,
                               zero_division=0), 2
        ),
        "F1-score (%)": round(
            100 * f1_score(y_true, pred, pos_label=1,
                            zero_division=0), 2
        )
    })

comparison = pd.DataFrame(rows)
display(comparison)

ensemble_pred = predictions["Ensemble"]
print("\nEnsemble classification report:")
print(classification_report(
    y_true, ensemble_pred,
    target_names=class_names,
    zero_division=0
))

cm = confusion_matrix(y_true, ensemble_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=class_names, yticklabels=class_names
)
plt.xlabel("Predicted class")
plt.ylabel("Actual class")
plt.title("VGG19 + ResNet50 + EfficientNetB2 Ensemble")
plt.show()
```

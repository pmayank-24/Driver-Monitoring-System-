```python

# =====================================================================
# DRIVER DROWSINESS DETECTION USING VGG19
# DATASET: GOOGLE DRIVE
# MODEL: VGG19 (IMAGENET PRETRAINED)
# EPOCHS: 30
# SINGLE TRAINING PHASE
# NO FINE-TUNING
# NO EARLY STOPPING
# =====================================================================


# =====================================================================
# 1. IMPORT LIBRARIES
# =====================================================================

import os
import re
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from collections import Counter

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from google.colab import drive


# =====================================================================
# 2. CONFIGURATION
# =====================================================================

SEED = 42

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 30

LEARNING_RATE = 0.0001

CLASS_NAMES = ["Awake", "Drowsy"]

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices("GPU"))


# =====================================================================
# 3. CONNECT GOOGLE DRIVE
# =====================================================================

print("\nConnecting Google Drive...")

drive.mount("/content/drive")


# =====================================================================
# 4. DATASET PATH
# =====================================================================

DATASET_ROOT = (
    "/content/drive/MyDrive/Driver_test/Driver_testing"
)

OUTPUT_PATH = (
    "/content/drive/MyDrive/Driver_test/VGG19_Results"
)

os.makedirs(OUTPUT_PATH, exist_ok=True)

# Verify dataset folders

for folder in ["train", "valid", "test"]:

    folder_path = os.path.join(
        DATASET_ROOT,
        folder
    )

    if not os.path.isdir(folder_path):

        raise FileNotFoundError(
            f"Dataset folder not found: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "images")
    ):

        raise FileNotFoundError(
            f"Images folder missing: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "labels")
    ):

        raise FileNotFoundError(
            f"Labels folder missing: {folder_path}"
        )

print("\nGoogle Drive connected successfully!")
print("Dataset Path:", DATASET_ROOT)
print("Output Path:", OUTPUT_PATH)


# =====================================================================
# 5. VERIFY CLASS MAPPING
# =====================================================================

print("\nChecking dataset configuration...")

DATA_YAML = os.path.join(
    DATASET_ROOT,
    "data.yaml"
)

if os.path.exists(DATA_YAML):

    with open(DATA_YAML, "r") as f:
        print(f.read())

print("\nClass 0 = Awake")
print("Class 1 = Drowsy")


# =====================================================================
# 6. LOAD IMAGES AND MATCH ANNOTATIONS
# FIXED FOR FILENAMES SUCH AS (1), (2), ETC.
# =====================================================================

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp"
)


def normalize_filename(filename):

    # Remove file extension

    name = os.path.splitext(filename)[0]

    # Remove duplicate suffix such as (1), (2), (3)

    name = re.sub(
        r"\s*\(\d+\)$",
        "",
        name
    )

    return name.lower()


def load_split(split_name):

    image_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "images"
    )

    label_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "labels"
    )

    image_paths = []
    image_labels = []

    missing_annotations = []
    empty_annotations = []

    # Build annotation lookup

    label_lookup = {}

    for label_file in os.listdir(label_dir):

        if not label_file.lower().endswith(".txt"):
            continue

        normalized = normalize_filename(
            label_file
        )

        label_lookup[normalized] = os.path.join(
            label_dir,
            label_file
        )

    # Read images

    for image_file in sorted(os.listdir(image_dir)):

        if not image_file.lower().endswith(
            IMAGE_EXTENSIONS
        ):
            continue

        image_path = os.path.join(
            image_dir,
            image_file
        )

        # First try exact filename matching

        exact_label = (
            os.path.splitext(image_file)[0] + ".txt"
        )

        exact_path = os.path.join(
            label_dir,
            exact_label
        )

        if os.path.exists(exact_path):

            label_path = exact_path

        else:

            # Fallback: normalize duplicate filename suffix

            normalized = normalize_filename(
                image_file
            )

            label_path = label_lookup.get(
                normalized
            )

        # Skip images with no matching annotation

        if label_path is None:

            missing_annotations.append(
                image_file
            )

            continue

        with open(label_path, "r") as f:

            lines = [
                line.strip()
                for line in f.readlines()
                if line.strip()
            ]

        if not lines:

            empty_annotations.append(
                image_file
            )

            continue

        class_ids = []

        for line in lines:

            values = line.split()

            try:

                class_id = int(
                    float(values[0])
                )

                if class_id not in [0, 1]:

                    raise ValueError(
                        f"Unexpected class ID: {class_id}"
                    )

                class_ids.append(class_id)

            except (ValueError, IndexError):

                raise ValueError(
                    f"Invalid annotation in {label_path}"
                )

        # Use majority class for image-level classification

        image_class = Counter(
            class_ids
        ).most_common(1)[0][0]

        image_paths.append(
            image_path
        )

        image_labels.append(
            image_class
        )

    print("\n====================================")
    print(split_name.upper(), "DATASET")
    print("====================================")

    print("Images loaded:", len(image_paths))

    print(
        "Missing annotations:",
        len(missing_annotations)
    )

    print(
        "Empty annotations:",
        len(empty_annotations)
    )

    if missing_annotations:

        print("\nFirst unmatched images:")

        for filename in missing_annotations[:10]:

            print(filename)

    if len(image_paths) == 0:

        raise ValueError(
            f"No labelled images found in {split_name}."
        )

    return (
        np.array(image_paths),
        np.array(image_labels, dtype=np.int32)
    )


# Load all dataset splits

train_paths, train_labels = load_split("train")

val_paths, val_labels = load_split("valid")

test_paths, test_labels = load_split("test")


# =====================================================================
# 7. DATASET SUMMARY
# =====================================================================

print("\n====================================")
print("FINAL DATASET SUMMARY")
print("====================================")


def show_distribution(name, labels):

    counts = Counter(labels.tolist())

    print(f"\n{name}")

    print("Total Images:", len(labels))

    print("Awake:", counts.get(0, 0))

    print("Drowsy:", counts.get(1, 0))


show_distribution(
    "TRAINING DATA",
    train_labels
)

show_distribution(
    "VALIDATION DATA",
    val_labels
)

show_distribution(
    "TESTING DATA",
    test_labels
)


if min(
    len(train_paths),
    len(val_paths),
    len(test_paths)
) == 0:

    raise ValueError(
        "One or more dataset splits are empty."
    )


# =====================================================================
# 8. IMAGE PREPROCESSING
# =====================================================================

def preprocess_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    label = tf.cast(
        label,
        tf.float32
    )

    return image, label


# =====================================================================
# 9. CREATE TENSORFLOW DATASETS
# =====================================================================

def create_dataset(
    paths,
    labels,
    training=False
):

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:

        dataset = dataset.shuffle(
            buffer_size=len(paths),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_dataset = create_dataset(
    train_paths,
    train_labels,
    training=True
)

val_dataset = create_dataset(
    val_paths,
    val_labels
)

test_dataset = create_dataset(
    test_paths,
    test_labels
)

print("\nTensorFlow datasets created successfully!")


# =====================================================================
# 10. LOAD PRETRAINED VGG19
# =====================================================================

print("\nLoading VGG19 ImageNet pretrained model...")

base_model = tf.keras.applications.VGG19(

    weights="imagenet",

    include_top=False,

    input_shape=(IMG_SIZE, IMG_SIZE, 3)

)

# Freeze all VGG19 layers

base_model.trainable = False

print("VGG19 loaded successfully!")


# =====================================================================
# 11. BUILD VGG19 CLASSIFICATION MODEL
# =====================================================================

inputs = tf.keras.Input(
    shape=(IMG_SIZE, IMG_SIZE, 3)
)

# VGG19 preprocessing

x = tf.keras.applications.vgg19.preprocess_input(
    inputs
)

# Feature extraction

x = base_model(
    x,
    training=False
)

# Global Average Pooling

x = tf.keras.layers.GlobalAveragePooling2D()(x)

# Fully Connected Layer

x = tf.keras.layers.Dense(
    512,
    activation="relu"
)(x)

x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dropout(0.5)(x)

# Second Dense Layer

x = tf.keras.layers.Dense(
    128,
    activation="relu"
)(x)

x = tf.keras.layers.Dropout(0.3)(x)

# Binary Classification

outputs = tf.keras.layers.Dense(
    1,
    activation="sigmoid"
)(x)

model = tf.keras.Model(
    inputs=inputs,
    outputs=outputs,
    name="VGG19_Driver_Drowsiness"
)

print("\nVGG19 MODEL SUMMARY")

model.summary()


# =====================================================================
# 12. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),

    loss="binary_crossentropy",

    metrics=[

        tf.keras.metrics.BinaryAccuracy(
            name="accuracy"
        ),

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]
)


# =====================================================================
# 13. CALLBACKS
# NO EARLY STOPPING
# =====================================================================

checkpoint_path = os.path.join(
    OUTPUT_PATH,
    "Best_VGG19_Model.keras"
)

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(

        filepath=checkpoint_path,

        monitor="val_loss",

        save_best_only=True,

        mode="min",

        verbose=1

    ),

    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=5,

        min_lr=1e-7,

        verbose=1

    )

]


# =====================================================================
# 14. TRAIN VGG19 MODEL
# EXACTLY 30 EPOCHS
# SINGLE TRAINING PHASE
# =====================================================================

print("\n====================================")
print("STARTING VGG19 TRAINING")
print("TOTAL EPOCHS: 30")
print("BATCH SIZE: 16")
print("SINGLE TRAINING PHASE")
print("====================================")

history = model.fit(

    train_dataset,

    validation_data=val_dataset,

    epochs=EPOCHS,

    callbacks=callbacks,

    verbose=1

)

print("\nVGG19 training completed!")


# =====================================================================
# 15. LOAD BEST VALIDATION MODEL
# =====================================================================

print("\nLoading best validation model...")

model = tf.keras.models.load_model(
    checkpoint_path
)


# =====================================================================
# 16. FINAL TEST EVALUATION
# =====================================================================

print("\n====================================")
print("FINAL TEST EVALUATION")
print("====================================")

test_results = model.evaluate(
    test_dataset,
    verbose=1
)

metric_names = [
    "Loss",
    "Accuracy",
    "Precision",
    "Recall",
    "AUC"
]

for name, value in zip(
    metric_names,
    test_results
):

    print(f"{name}: {value:.4f}")


# =====================================================================
# 17. GENERATE PREDICTIONS
# =====================================================================

y_prob = model.predict(
    test_dataset,
    verbose=1
).flatten()

y_pred = (
    y_prob >= 0.5
).astype(int)

y_true = test_labels


# =====================================================================
# 18. CLASSIFICATION REPORT
# =====================================================================

print("\n====================================")
print("CLASSIFICATION REPORT")
print("====================================")

print(

    classification_report(

        y_true,

        y_pred,

        labels=[0, 1],

        target_names=CLASS_NAMES,

        digits=4,

        zero_division=0

    )

)


# =====================================================================
# 19. PERFORMANCE METRICS
# =====================================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

auc = roc_auc_score(
    y_true,
    y_prob
)

print("\n====================================")
print("VGG19 FINAL PERFORMANCE")
print("====================================")

print(f"Accuracy  : {accuracy * 100:.2f}%")

print(f"Precision : {precision * 100:.2f}%")

print(f"Recall    : {recall * 100:.2f}%")

print(f"F1 Score  : {f1 * 100:.2f}%")

print(f"ROC-AUC   : {auc * 100:.2f}%")


# =====================================================================
# 20. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=CLASS_NAMES

)

disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)

plt.title(
    "VGG19 - Confusion Matrix"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Confusion_Matrix.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 21. TRAINING AND VALIDATION ACCURACY GRAPH
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "VGG19 Training and Validation Accuracy"
)

plt.xlabel("Epochs")

plt.ylabel("Accuracy")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Accuracy_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 22. TRAINING AND VALIDATION LOSS GRAPH
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "VGG19 Training and Validation Loss"
)

plt.xlabel("Epochs")

plt.ylabel("Loss")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Loss_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 23. SAVE TRAINING HISTORY
# =====================================================================

history_df = pd.DataFrame(
    history.history
)

history_df.insert(
    0,
    "Epoch",
    range(1, EPOCHS + 1)
)

history_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "Training_History.csv"
    ),
    index=False
)


# =====================================================================
# 24. SAVE FINAL MODEL
# =====================================================================

final_model_path = os.path.join(
    OUTPUT_PATH,
    "VGG19_Driver_Drowsiness_Final.keras"
)

model.save(final_model_path)

print("\nFinal model saved successfully!")


# =====================================================================
# 25. SAVE PERFORMANCE RESULTS
# =====================================================================

results_df = pd.DataFrame({

    "Model": ["VGG19"],

    "Epochs": [EPOCHS],

    "Batch_Size": [BATCH_SIZE],

    "Image_Size": ["224x224"],

    "Accuracy": [accuracy * 100],

    "Precision": [precision * 100],

    "Recall": [recall * 100],

    "F1_Score": [f1 * 100],

    "ROC_AUC": [auc * 100]

})

results_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "VGG19_Performance_Results.csv"
    ),
    index=False
)

print("\n====================================")
print("FINAL PERFORMANCE TABLE")
print("====================================")

print(
    results_df.to_string(index=False)
)


# =====================================================================
# 26. INDIVIDUAL IMAGE PREDICTION
# =====================================================================

def predict_drowsiness(image_path):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    image_batch = tf.expand_dims(
        image,
        axis=0
    )

    probability = float(
        model.predict(
            image_batch,
            verbose=0
        )[0][0]
    )

    predicted_class = (
        "Drowsy"
        if probability >= 0.5
        else "Awake"
    )

    confidence = (
        probability
        if probability >= 0.5
        else 1 - probability
    )

    print("\nPrediction:", predicted_class)

    print(
        f"Confidence: {confidence * 100:.2f}%"
    )

    plt.figure(
        figsize=(5, 5)
    )

    plt.imshow(
        tf.cast(image, tf.uint8)
    )

    plt.title(
        f"{predicted_class} - {confidence * 100:.2f}%"
    )

    plt.axis("off")

    plt.show()


# =====================================================================
# 27. FINAL SUMMARY
# =====================================================================

print("\n====================================")
print("VGG19 TRAINING COMPLETED!")
print("====================================")

print("Model: VGG19")

print("Total Epochs:", EPOCHS)

print("Batch Size:", BATCH_SIZE)

print("Training Phase: Single")

print("Fine-Tuning: No")

print("Early Stopping: No")

print(
    f"Final Test Accuracy: {accuracy * 100:.2f}%"
)

print("\nAll files saved in Google Drive:")

print(OUTPUT_PATH)

print("\nSaved Files:")

print("1. Best_VGG19_Model.keras")

print("2. VGG19_Driver_Drowsiness_Final.keras")

print("3. VGG19_Performance_Results.csv")

print("4. Training_History.csv")

print("5. Confusion_Matrix.png")

print("6. Accuracy_Graph.png")

print("7. Loss_Graph.png")

print("\nALL TASKS COMPLETED SUCCESSFULLY!")

# =====================================================================
# END OF VGG19 PROJECT
# =====================================================================
```

```python

# =====================================================================
# DRIVER DROWSINESS DETECTION USING EFFICIENTNETB0
# GOOGLE DRIVE DATASET
# 30 EPOCHS - SINGLE TRAINING PHASE
# NO FINE-TUNING
# NO EARLY STOPPING
# =====================================================================


# =====================================================================
# 1. IMPORT LIBRARIES
# =====================================================================

import os
import re
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from collections import Counter

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from google.colab import drive


# =====================================================================
# 2. CONFIGURATION
# =====================================================================

SEED = 42

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 30

LEARNING_RATE = 0.0001

CLASS_NAMES = ["Awake", "Drowsy"]

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices("GPU"))


# =====================================================================
# 3. CONNECT GOOGLE DRIVE
# =====================================================================

print("\nConnecting Google Drive...")

drive.mount("/content/drive")


# =====================================================================
# 4. DATASET PATH
# =====================================================================

DATASET_ROOT = (
    "/content/drive/MyDrive/Driver_test/Driver_testing"
)

OUTPUT_PATH = (
    "/content/drive/MyDrive/Driver_test/EfficientNetB0_Results"
)

os.makedirs(OUTPUT_PATH, exist_ok=True)

# Verify dataset folders

for folder in ["train", "valid", "test"]:

    folder_path = os.path.join(
        DATASET_ROOT,
        folder
    )

    if not os.path.isdir(folder_path):
        raise FileNotFoundError(
            f"Dataset folder not found: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "images")
    ):
        raise FileNotFoundError(
            f"Images folder missing: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "labels")
    ):
        raise FileNotFoundError(
            f"Labels folder missing: {folder_path}"
        )

print("\nGoogle Drive connected successfully!")
print("Dataset Path:", DATASET_ROOT)
print("Output Path:", OUTPUT_PATH)


# =====================================================================
# 5. CHECK DATASET CONFIGURATION
# =====================================================================

DATA_YAML = os.path.join(
    DATASET_ROOT,
    "data.yaml"
)

if os.path.exists(DATA_YAML):

    print("\nDataset Configuration:")

    with open(DATA_YAML, "r") as f:
        print(f.read())

print("\nClass 0 = Awake")
print("Class 1 = Drowsy")


# =====================================================================
# 6. LOAD IMAGES AND MATCH YOLO ANNOTATIONS
# HANDLES FILENAMES WITH (1), (2), ETC.
# =====================================================================

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp"
)


def normalize_filename(filename):

    name = os.path.splitext(filename)[0]

    name = re.sub(
        r"\s*\(\d+\)$",
        "",
        name
    )

    return name.lower()


def load_split(split_name):

    image_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "images"
    )

    label_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "labels"
    )

    image_paths = []
    image_labels = []

    missing_annotations = []
    empty_annotations = []

    # Create annotation lookup

    label_lookup = {}

    for label_file in sorted(os.listdir(label_dir)):

        if not label_file.lower().endswith(".txt"):
            continue

        normalized = normalize_filename(
            label_file
        )

        label_lookup[normalized] = os.path.join(
            label_dir,
            label_file
        )

    # Read images

    for image_file in sorted(os.listdir(image_dir)):

        if not image_file.lower().endswith(
            IMAGE_EXTENSIONS
        ):
            continue

        image_path = os.path.join(
            image_dir,
            image_file
        )

        # First check exact filename

        exact_label = (
            os.path.splitext(image_file)[0] + ".txt"
        )

        exact_path = os.path.join(
            label_dir,
            exact_label
        )

        if os.path.exists(exact_path):

            label_path = exact_path

        else:

            # Try matching normalized filename

            normalized = normalize_filename(
                image_file
            )

            label_path = label_lookup.get(
                normalized
            )

        if label_path is None:

            missing_annotations.append(
                image_file
            )

            continue

        with open(label_path, "r") as f:

            lines = [
                line.strip()
                for line in f.readlines()
                if line.strip()
            ]

        if not lines:

            empty_annotations.append(
                image_file
            )

            continue

        class_ids = []

        for line in lines:

            values = line.split()

            try:

                class_id = int(
                    float(values[0])
                )

                if class_id not in [0, 1]:

                    raise ValueError(
                        f"Unexpected class ID: {class_id}"
                    )

                class_ids.append(class_id)

            except (ValueError, IndexError):

                raise ValueError(
                    f"Invalid annotation: {label_path}"
                )

        # Majority class for image-level classification

        image_class = Counter(
            class_ids
        ).most_common(1)[0][0]

        image_paths.append(
            image_path
        )

        image_labels.append(
            image_class
        )

    print("\n====================================")
    print(split_name.upper(), "DATASET")
    print("====================================")

    print("Images loaded:", len(image_paths))

    print(
        "Missing annotations:",
        len(missing_annotations)
    )

    print(
        "Empty annotations:",
        len(empty_annotations)
    )

    if missing_annotations:

        print("\nFirst unmatched images:")

        for filename in missing_annotations[:10]:
            print(filename)

    if len(image_paths) == 0:

        raise ValueError(
            f"No labelled images found in {split_name}."
        )

    return (
        np.array(image_paths),
        np.array(image_labels, dtype=np.int32)
    )


# Load dataset

train_paths, train_labels = load_split("train")

val_paths, val_labels = load_split("valid")

test_paths, test_labels = load_split("test")


# =====================================================================
# 7. DATASET SUMMARY
# =====================================================================

print("\n====================================")
print("DATASET SUMMARY")
print("====================================")


def show_distribution(name, labels):

    counts = Counter(labels.tolist())

    print(f"\n{name}")

    print("Total Images:", len(labels))

    print("Awake:", counts.get(0, 0))

    print("Drowsy:", counts.get(1, 0))


show_distribution(
    "TRAINING DATA",
    train_labels
)

show_distribution(
    "VALIDATION DATA",
    val_labels
)

show_distribution(
    "TESTING DATA",
    test_labels
)


if min(
    len(train_paths),
    len(val_paths),
    len(test_paths)
) == 0:

    raise ValueError(
        "One or more dataset splits are empty."
    )


# =====================================================================
# 8. IMAGE PREPROCESSING
# =====================================================================

def preprocess_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    # EfficientNetB0 expects image values in 0-255.
    # Its internal preprocessing handles rescaling.

    image = tf.cast(
        image,
        tf.float32
    )

    label = tf.cast(
        label,
        tf.float32
    )

    return image, label


# =====================================================================
# 9. CREATE TENSORFLOW DATASETS
# =====================================================================

def create_dataset(
    paths,
    labels,
    training=False
):

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:

        dataset = dataset.shuffle(
            buffer_size=len(paths),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_dataset = create_dataset(
    train_paths,
    train_labels,
    training=True
)

val_dataset = create_dataset(
    val_paths,
    val_labels
)

test_dataset = create_dataset(
    test_paths,
    test_labels
)

print("\nTensorFlow datasets created successfully!")


# =====================================================================
# 10. LOAD PRETRAINED EFFICIENTNETB0
# =====================================================================

print("\nLoading EfficientNetB0 ImageNet pretrained model...")

base_model = tf.keras.applications.EfficientNetB0(

    weights="imagenet",

    include_top=False,

    input_shape=(IMG_SIZE, IMG_SIZE, 3),

    pooling="avg"

)

# Freeze all EfficientNetB0 layers

base_model.trainable = False

print("EfficientNetB0 loaded successfully!")


# =====================================================================
# 11. BUILD EFFICIENTNETB0 CLASSIFICATION MODEL
# =====================================================================

inputs = tf.keras.Input(
    shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Feature extraction

x = base_model(
    inputs,
    training=False
)

# Fully connected layer

x = tf.keras.layers.Dense(
    256,
    activation="relu"
)(x)

x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dropout(0.4)(x)

# Second fully connected layer

x = tf.keras.layers.Dense(
    128,
    activation="relu"
)(x)

x = tf.keras.layers.Dropout(0.3)(x)

# Binary classification output

outputs = tf.keras.layers.Dense(
    1,
    activation="sigmoid"
)(x)

model = tf.keras.Model(
    inputs=inputs,
    outputs=outputs,
    name="EfficientNetB0_Driver_Drowsiness"
)

print("\nEfficientNetB0 Model Summary:")

model.summary()


# =====================================================================
# 12. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),

    loss="binary_crossentropy",

    metrics=[

        tf.keras.metrics.BinaryAccuracy(
            name="accuracy"
        ),

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]
)


# =====================================================================
# 13. CALLBACKS
# NO EARLY STOPPING
# =====================================================================

checkpoint_path = os.path.join(
    OUTPUT_PATH,
    "Best_EfficientNetB0_Model.keras"
)

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(

        filepath=checkpoint_path,

        monitor="val_loss",

        save_best_only=True,

        mode="min",

        verbose=1

    ),

    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=5,

        min_lr=1e-7,

        verbose=1

    )

]


# =====================================================================
# 14. TRAIN EFFICIENTNETB0
# EXACTLY 30 EPOCHS
# SINGLE TRAINING PHASE
# =====================================================================

print("\n====================================")
print("STARTING EFFICIENTNETB0 TRAINING")
print("TOTAL EPOCHS: 30")
print("BATCH SIZE: 16")
print("SINGLE TRAINING PHASE")
print("====================================")

history = model.fit(

    train_dataset,

    validation_data=val_dataset,

    epochs=EPOCHS,

    callbacks=callbacks,

    verbose=1

)

print("\nEfficientNetB0 training completed!")


# =====================================================================
# 15. LOAD BEST VALIDATION MODEL
# =====================================================================

print("\nLoading best validation model...")

model = tf.keras.models.load_model(
    checkpoint_path
)


# =====================================================================
# 16. FINAL TEST EVALUATION
# =====================================================================

print("\n====================================")
print("FINAL TEST EVALUATION")
print("====================================")

test_results = model.evaluate(
    test_dataset,
    verbose=1
)

metric_names = [
    "Loss",
    "Accuracy",
    "Precision",
    "Recall",
    "AUC"
]

for name, value in zip(
    metric_names,
    test_results
):

    print(f"{name}: {value:.4f}")


# =====================================================================
# 17. GENERATE PREDICTIONS
# =====================================================================

y_prob = model.predict(
    test_dataset,
    verbose=1
).flatten()

y_pred = (
    y_prob >= 0.5
).astype(int)

y_true = test_labels


# =====================================================================
# 18. CLASSIFICATION REPORT
# =====================================================================

print("\n====================================")
print("CLASSIFICATION REPORT")
print("====================================")

print(

    classification_report(

        y_true,

        y_pred,

        labels=[0, 1],

        target_names=CLASS_NAMES,

        digits=4,

        zero_division=0

    )

)


# =====================================================================
# 19. PERFORMANCE METRICS
# =====================================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

auc = roc_auc_score(
    y_true,
    y_prob
)

print("\n====================================")
print("EFFICIENTNETB0 FINAL PERFORMANCE")
print("====================================")

print(f"Accuracy  : {accuracy * 100:.2f}%")

print(f"Precision : {precision * 100:.2f}%")

print(f"Recall    : {recall * 100:.2f}%")

print(f"F1 Score  : {f1 * 100:.2f}%")

print(f"ROC-AUC   : {auc * 100:.2f}%")


# =====================================================================
# 20. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=CLASS_NAMES

)

disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)

plt.title(
    "EfficientNetB0 - Confusion Matrix"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Confusion_Matrix.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 21. TRAINING AND VALIDATION ACCURACY
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "EfficientNetB0 Training and Validation Accuracy"
)

plt.xlabel("Epochs")

plt.ylabel("Accuracy")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Accuracy_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 22. TRAINING AND VALIDATION LOSS
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "EfficientNetB0 Training and Validation Loss"
)

plt.xlabel("Epochs")

plt.ylabel("Loss")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Loss_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 23. SAVE TRAINING HISTORY
# =====================================================================

history_df = pd.DataFrame(
    history.history
)

history_df.insert(
    0,
    "Epoch",
    range(1, EPOCHS + 1)
)

history_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "Training_History.csv"
    ),
    index=False
)


# =====================================================================
# 24. SAVE FINAL MODEL
# =====================================================================

final_model_path = os.path.join(
    OUTPUT_PATH,
    "EfficientNetB0_Driver_Drowsiness_Final.keras"
)

model.save(final_model_path)

print("\nFinal model saved successfully!")


# =====================================================================
# 25. SAVE PERFORMANCE RESULTS
# =====================================================================

results_df = pd.DataFrame({

    "Model": ["EfficientNetB0"],

    "Epochs": [EPOCHS],

    "Batch_Size": [BATCH_SIZE],

    "Image_Size": ["224x224"],

    "Accuracy": [accuracy * 100],

    "Precision": [precision * 100],

    "Recall": [recall * 100],

    "F1_Score": [f1 * 100],

    "ROC_AUC": [auc * 100]

})

results_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "EfficientNetB0_Performance_Results.csv"
    ),
    index=False
)

print("\n====================================")
print("FINAL PERFORMANCE TABLE")
print("====================================")

print(
    results_df.to_string(index=False)
)


# =====================================================================
# 26. INDIVIDUAL IMAGE PREDICTION
# =====================================================================

def predict_drowsiness(image_path):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    image_batch = tf.expand_dims(
        image,
        axis=0
    )

    probability = float(
        model.predict(
            image_batch,
            verbose=0
        )[0][0]
    )

    predicted_class = (
        "Drowsy"
        if probability >= 0.5
        else "Awake"
    )

    confidence = (
        probability
        if probability >= 0.5
        else 1 - probability
    )

    print("\nPrediction:", predicted_class)

    print(
        f"Confidence: {confidence * 100:.2f}%"
    )

    plt.figure(
        figsize=(5, 5)
    )

    plt.imshow(
        tf.cast(image, tf.uint8)
    )

    plt.title(
        f"{predicted_class} - {confidence * 100:.2f}%"
    )

    plt.axis("off")

    plt.show()


# =====================================================================
# 27. FINAL SUMMARY
# =====================================================================

print("\n====================================")
print("EFFICIENTNETB0 TRAINING COMPLETED!")
print("====================================")

print("Model: EfficientNetB0")

print("Total Epochs:", EPOCHS)

print("Batch Size:", BATCH_SIZE)

print("Training Phase: Single")

print("Fine-Tuning: No")

print("Early Stopping: No")

print(
    f"Final Test Accuracy: {accuracy * 100:.2f}%"
)

print("\nAll files saved in Google Drive:")

print(OUTPUT_PATH)

print("\nSaved Files:")

print("1. Best_EfficientNetB0_Model.keras")

print("2. EfficientNetB0_Driver_Drowsiness_Final.keras")

print("3. EfficientNetB0_Performance_Results.csv")

print("4. Training_History.csv")

print("5. Confusion_Matrix.png")

print("6. Accuracy_Graph.png")

print("7. Loss_Graph.png")

print("\nALL TASKS COMPLETED SUCCESSFULLY!")

# =====================================================================
# END OF EFFICIENTNETB0 PROJECT
# =====================================================================
```

```python

# =====================================================================
# DRIVER DROWSINESS DETECTION USING RESNET50
# GOOGLE DRIVE DATASET
# IMAGENET PRETRAINED MODEL
# 30 EPOCHS - SINGLE TRAINING PHASE
# NO FINE-TUNING
# NO EARLY STOPPING
# =====================================================================


# =====================================================================
# 1. IMPORT LIBRARIES
# =====================================================================

import os
import re
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from collections import Counter

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from google.colab import drive


# =====================================================================
# 2. CONFIGURATION
# =====================================================================

SEED = 42

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 30

LEARNING_RATE = 0.0001

CLASS_NAMES = ["Awake", "Drowsy"]

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices("GPU"))


# =====================================================================
# 3. CONNECT GOOGLE DRIVE
# =====================================================================

print("\nConnecting Google Drive...")

drive.mount("/content/drive")


# =====================================================================
# 4. DATASET PATH
# =====================================================================

DATASET_ROOT = (
    "/content/drive/MyDrive/Driver_test/Driver_testing"
)

OUTPUT_PATH = (
    "/content/drive/MyDrive/Driver_test/ResNet50_Results"
)

os.makedirs(OUTPUT_PATH, exist_ok=True)

# Verify dataset folders

for folder in ["train", "valid", "test"]:

    folder_path = os.path.join(
        DATASET_ROOT,
        folder
    )

    if not os.path.isdir(folder_path):
        raise FileNotFoundError(
            f"Dataset folder not found: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "images")
    ):
        raise FileNotFoundError(
            f"Images folder missing: {folder_path}"
        )

    if not os.path.isdir(
        os.path.join(folder_path, "labels")
    ):
        raise FileNotFoundError(
            f"Labels folder missing: {folder_path}"
        )

print("\nGoogle Drive connected successfully!")
print("Dataset Path:", DATASET_ROOT)
print("Output Path:", OUTPUT_PATH)


# =====================================================================
# 5. CHECK DATASET CONFIGURATION
# =====================================================================

DATA_YAML = os.path.join(
    DATASET_ROOT,
    "data.yaml"
)

if os.path.exists(DATA_YAML):

    print("\nDataset Configuration:")

    with open(DATA_YAML, "r") as f:
        print(f.read())

print("\nClass 0 = Awake")
print("Class 1 = Drowsy")


# =====================================================================
# 6. LOAD IMAGES AND MATCH YOLO ANNOTATIONS
# HANDLES FILENAMES WITH (1), (2), ETC.
# =====================================================================

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp"
)


def normalize_filename(filename):

    name = os.path.splitext(filename)[0]

    name = re.sub(
        r"\s*\(\d+\)$",
        "",
        name
    )

    return name.lower()


def load_split(split_name):

    image_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "images"
    )

    label_dir = os.path.join(
        DATASET_ROOT,
        split_name,
        "labels"
    )

    image_paths = []
    image_labels = []

    missing_annotations = []
    empty_annotations = []

    # Build annotation lookup

    label_lookup = {}

    for label_file in sorted(os.listdir(label_dir)):

        if not label_file.lower().endswith(".txt"):
            continue

        normalized = normalize_filename(
            label_file
        )

        label_lookup[normalized] = os.path.join(
            label_dir,
            label_file
        )

    # Load images

    for image_file in sorted(os.listdir(image_dir)):

        if not image_file.lower().endswith(
            IMAGE_EXTENSIONS
        ):
            continue

        image_path = os.path.join(
            image_dir,
            image_file
        )

        # First check exact filename

        exact_label = (
            os.path.splitext(image_file)[0] + ".txt"
        )

        exact_path = os.path.join(
            label_dir,
            exact_label
        )

        if os.path.exists(exact_path):

            label_path = exact_path

        else:

            # Try normalized filename matching

            normalized = normalize_filename(
                image_file
            )

            label_path = label_lookup.get(
                normalized
            )

        if label_path is None:

            missing_annotations.append(
                image_file
            )

            continue

        with open(label_path, "r") as f:

            lines = [
                line.strip()
                for line in f.readlines()
                if line.strip()
            ]

        if not lines:

            empty_annotations.append(
                image_file
            )

            continue

        class_ids = []

        for line in lines:

            values = line.split()

            try:

                class_id = int(
                    float(values[0])
                )

                if class_id not in [0, 1]:

                    raise ValueError(
                        f"Unexpected class ID: {class_id}"
                    )

                class_ids.append(class_id)

            except (ValueError, IndexError):

                raise ValueError(
                    f"Invalid annotation: {label_path}"
                )

        # Image-level majority class

        image_class = Counter(
            class_ids
        ).most_common(1)[0][0]

        image_paths.append(
            image_path
        )

        image_labels.append(
            image_class
        )

    print("\n====================================")
    print(split_name.upper(), "DATASET")
    print("====================================")

    print("Images loaded:", len(image_paths))

    print(
        "Missing annotations:",
        len(missing_annotations)
    )

    print(
        "Empty annotations:",
        len(empty_annotations)
    )

    if missing_annotations:

        print("\nFirst unmatched images:")

        for filename in missing_annotations[:10]:
            print(filename)

    if len(image_paths) == 0:

        raise ValueError(
            f"No labelled images found in {split_name}."
        )

    return (
        np.array(image_paths),
        np.array(image_labels, dtype=np.int32)
    )


# Load all dataset splits

train_paths, train_labels = load_split("train")

val_paths, val_labels = load_split("valid")

test_paths, test_labels = load_split("test")


# =====================================================================
# 7. DATASET SUMMARY
# =====================================================================

print("\n====================================")
print("DATASET SUMMARY")
print("====================================")


def show_distribution(name, labels):

    counts = Counter(labels.tolist())

    print(f"\n{name}")

    print("Total Images:", len(labels))

    print("Awake:", counts.get(0, 0))

    print("Drowsy:", counts.get(1, 0))


show_distribution(
    "TRAINING DATA",
    train_labels
)

show_distribution(
    "VALIDATION DATA",
    val_labels
)

show_distribution(
    "TESTING DATA",
    test_labels
)


if min(
    len(train_paths),
    len(val_paths),
    len(test_paths)
) == 0:

    raise ValueError(
        "One or more dataset splits are empty."
    )


# =====================================================================
# 8. IMAGE PREPROCESSING
# RESNET50 SPECIFIC PREPROCESSING
# =====================================================================

def preprocess_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    # ResNet50 preprocessing

    image = tf.keras.applications.resnet50.preprocess_input(
        image
    )

    label = tf.cast(
        label,
        tf.float32
    )

    return image, label


# =====================================================================
# 9. CREATE TENSORFLOW DATASETS
# =====================================================================

def create_dataset(
    paths,
    labels,
    training=False
):

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:

        dataset = dataset.shuffle(
            buffer_size=len(paths),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        preprocess_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_dataset = create_dataset(
    train_paths,
    train_labels,
    training=True
)

val_dataset = create_dataset(
    val_paths,
    val_labels
)

test_dataset = create_dataset(
    test_paths,
    test_labels
)

print("\nTensorFlow datasets created successfully!")


# =====================================================================
# 10. LOAD PRETRAINED RESNET50
# =====================================================================

print("\nLoading ResNet50 ImageNet pretrained model...")

base_model = tf.keras.applications.ResNet50(

    weights="imagenet",

    include_top=False,

    input_shape=(IMG_SIZE, IMG_SIZE, 3),

    pooling="avg"

)

# Freeze all ResNet50 layers

base_model.trainable = False

print("ResNet50 loaded successfully!")


# =====================================================================
# 11. BUILD RESNET50 CLASSIFICATION MODEL
# =====================================================================

inputs = tf.keras.Input(
    shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Feature extraction

x = base_model(
    inputs,
    training=False
)

# Fully connected layer

x = tf.keras.layers.Dense(
    256,
    activation="relu"
)(x)

x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dropout(0.4)(x)

# Second fully connected layer

x = tf.keras.layers.Dense(
    128,
    activation="relu"
)(x)

x = tf.keras.layers.Dropout(0.3)(x)

# Binary classification output

outputs = tf.keras.layers.Dense(
    1,
    activation="sigmoid"
)(x)

model = tf.keras.Model(
    inputs=inputs,
    outputs=outputs,
    name="ResNet50_Driver_Drowsiness"
)

print("\nResNet50 Model Summary:")

model.summary()


# =====================================================================
# 12. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),

    loss="binary_crossentropy",

    metrics=[

        tf.keras.metrics.BinaryAccuracy(
            name="accuracy"
        ),

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]
)


# =====================================================================
# 13. CALLBACKS
# NO EARLY STOPPING
# =====================================================================

checkpoint_path = os.path.join(
    OUTPUT_PATH,
    "Best_ResNet50_Model.keras"
)

callbacks = [

    tf.keras.callbacks.ModelCheckpoint(

        filepath=checkpoint_path,

        monitor="val_loss",

        save_best_only=True,

        mode="min",

        verbose=1

    ),

    tf.keras.callbacks.ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.3,

        patience=5,

        min_lr=1e-7,

        verbose=1

    )

]


# =====================================================================
# 14. TRAIN RESNET50
# EXACTLY 30 EPOCHS
# SINGLE TRAINING PHASE
# =====================================================================

print("\n====================================")
print("STARTING RESNET50 TRAINING")
print("TOTAL EPOCHS: 30")
print("BATCH SIZE: 16")
print("SINGLE TRAINING PHASE")
print("====================================")

history = model.fit(

    train_dataset,

    validation_data=val_dataset,

    epochs=EPOCHS,

    callbacks=callbacks,

    verbose=1

)

print("\nResNet50 training completed!")


# =====================================================================
# 15. LOAD BEST VALIDATION MODEL
# =====================================================================

print("\nLoading best validation model...")

model = tf.keras.models.load_model(
    checkpoint_path
)


# =====================================================================
# 16. FINAL TEST EVALUATION
# =====================================================================

print("\n====================================")
print("FINAL TEST EVALUATION")
print("====================================")

test_results = model.evaluate(
    test_dataset,
    verbose=1
)

metric_names = [
    "Loss",
    "Accuracy",
    "Precision",
    "Recall",
    "AUC"
]

for name, value in zip(
    metric_names,
    test_results
):

    print(f"{name}: {value:.4f}")


# =====================================================================
# 17. GENERATE PREDICTIONS
# =====================================================================

y_prob = model.predict(
    test_dataset,
    verbose=1
).flatten()

y_pred = (
    y_prob >= 0.5
).astype(int)

y_true = test_labels


# =====================================================================
# 18. CLASSIFICATION REPORT
# =====================================================================

print("\n====================================")
print("CLASSIFICATION REPORT")
print("====================================")

print(

    classification_report(

        y_true,

        y_pred,

        labels=[0, 1],

        target_names=CLASS_NAMES,

        digits=4,

        zero_division=0

    )

)


# =====================================================================
# 19. PERFORMANCE METRICS
# =====================================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

auc = roc_auc_score(
    y_true,
    y_prob
)

print("\n====================================")
print("RESNET50 FINAL PERFORMANCE")
print("====================================")

print(f"Accuracy  : {accuracy * 100:.2f}%")

print(f"Precision : {precision * 100:.2f}%")

print(f"Recall    : {recall * 100:.2f}%")

print(f"F1 Score  : {f1 * 100:.2f}%")

print(f"ROC-AUC   : {auc * 100:.2f}%")


# =====================================================================
# 20. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=CLASS_NAMES

)

disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)

plt.title(
    "ResNet50 - Confusion Matrix"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Confusion_Matrix.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 21. TRAINING AND VALIDATION ACCURACY
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "ResNet50 Training and Validation Accuracy"
)

plt.xlabel("Epochs")

plt.ylabel("Accuracy")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Accuracy_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 22. TRAINING AND VALIDATION LOSS
# =====================================================================

plt.figure(
    figsize=(10, 5)
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    range(1, EPOCHS + 1),
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "ResNet50 Training and Validation Loss"
)

plt.xlabel("Epochs")

plt.ylabel("Loss")

plt.xticks(
    range(1, EPOCHS + 1, 2)
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "Loss_Graph.png"
    ),
    dpi=300
)

plt.show()


# =====================================================================
# 23. SAVE TRAINING HISTORY
# =====================================================================

history_df = pd.DataFrame(
    history.history
)

history_df.insert(
    0,
    "Epoch",
    range(1, EPOCHS + 1)
)

history_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "Training_History.csv"
    ),
    index=False
)


# =====================================================================
# 24. SAVE FINAL MODEL
# =====================================================================

final_model_path = os.path.join(
    OUTPUT_PATH,
    "ResNet50_Driver_Drowsiness_Final.keras"
)

model.save(final_model_path)

print("\nFinal model saved successfully!")


# =====================================================================
# 25. SAVE PERFORMANCE RESULTS
# =====================================================================

results_df = pd.DataFrame({

    "Model": ["ResNet50"],

    "Epochs": [EPOCHS],

    "Batch_Size": [BATCH_SIZE],

    "Image_Size": ["224x224"],

    "Accuracy": [accuracy * 100],

    "Precision": [precision * 100],

    "Recall": [recall * 100],

    "F1_Score": [f1 * 100],

    "ROC_AUC": [auc * 100]

})

results_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "ResNet50_Performance_Results.csv"
    ),
    index=False
)

print("\n====================================")
print("FINAL PERFORMANCE TABLE")
print("====================================")

print(
    results_df.to_string(index=False)
)


# =====================================================================
# 26. INDIVIDUAL IMAGE PREDICTION
# =====================================================================

def predict_drowsiness(image_path):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape(
        [None, None, 3]
    )

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    # ResNet50 preprocessing

    image = tf.keras.applications.resnet50.preprocess_input(
        image
    )

    image_batch = tf.expand_dims(
        image,
        axis=0
    )

    probability = float(
        model.predict(
            image_batch,
            verbose=0
        )[0][0]
    )

    predicted_class = (
        "Drowsy"
        if probability >= 0.5
        else "Awake"
    )

    confidence = (
        probability
        if probability >= 0.5
        else 1 - probability
    )

    print("\nPrediction:", predicted_class)

    print(
        f"Confidence: {confidence * 100:.2f}%"
    )

    # Display original image

    original_image = tf.io.decode_image(
        tf.io.read_file(image_path),
        channels=3
    )

    original_image = tf.image.resize(
        original_image,
        [IMG_SIZE, IMG_SIZE]
    )

    plt.figure(figsize=(5, 5))

    plt.imshow(
        tf.cast(original_image, tf.uint8)
    )

    plt.title(
        f"{predicted_class} - {confidence * 100:.2f}%"
    )

    plt.axis("off")

    plt.show()


# =====================================================================
# 27. FINAL SUMMARY
# =====================================================================

print("\n====================================")
print("RESNET50 TRAINING COMPLETED!")
print("====================================")

print("Model: ResNet50")

print("Total Epochs:", EPOCHS)

print("Batch Size:", BATCH_SIZE)

print("Training Phase: Single")

print("Fine-Tuning: No")

print("Early Stopping: No")

print(
    f"Final Test Accuracy: {accuracy * 100:.2f}%"
)

print("\nAll files saved in Google Drive:")

print(OUTPUT_PATH)

print("\nSaved Files:")

print("1. Best_ResNet50_Model.keras")

print("2. ResNet50_Driver_Drowsiness_Final.keras")

print("3. ResNet50_Performance_Results.csv")

print("4. Training_History.csv")

print("5. Confusion_Matrix.png")

print("6. Accuracy_Graph.png")

print("7. Loss_Graph.png")

print("\nALL TASKS COMPLETED SUCCESSFULLY!")

# =====================================================================
# END OF RESNET50 PROJECT
# =====================================================================
```

```python
# ============================================================
# DRIVER DROWSINESS DETECTION USING DENSENET121
# Google Colab - Complete Single Cell Code
# 30 Epochs | No Fine-Tuning | No Early Stopping
# ============================================================

# 1. IMPORT LIBRARIES
import os
import re
import glob
import json
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from google.colab import drive
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.applications.densenet import preprocess_input
from tensorflow.keras.layers import (
    Input, GlobalAveragePooling2D, Dense,
    Dropout, BatchNormalization
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import (
    ModelCheckpoint, ReduceLROnPlateau, CSVLogger
)
from sklearn.metrics import (
    classification_report, confusion_matrix,
    accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

# 2. REPRODUCIBILITY
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# 3. MOUNT GOOGLE DRIVE
drive.mount('/content/drive')

# 4. CONFIGURATION
DATASET_PATH = "/content/drive/MyDrive/Driver_test/Driver_testing"
OUTPUT_PATH = "/content/drive/MyDrive/Driver_test/DenseNet121_Results"

IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS = 30
NUM_CLASSES = 2

os.makedirs(OUTPUT_PATH, exist_ok=True)

print("TensorFlow version:", tf.__version__)
print("Dataset path:", DATASET_PATH)
print("Output path:", OUTPUT_PATH)

# 5. READ DATA.YAML
yaml_path = os.path.join(DATASET_PATH, "data.yaml")

if not os.path.exists(yaml_path):
    raise FileNotFoundError(
        "data.yaml not found. Check your dataset folder."
    )

try:
    import yaml
    with open(yaml_path, "r") as f:
        yaml_data = yaml.safe_load(f)

    class_names = yaml_data.get("names", ["Awake", "Drowsy"])

    if isinstance(class_names, dict):
        class_names = [
            class_names[k] for k in sorted(class_names.keys())
        ]

    print("\nClass names from data.yaml:", class_names)

except Exception as e:
    print("Could not read class names:", e)
    class_names = ["Awake", "Drowsy"]

if len(class_names) != 2:
    raise ValueError(
        f"Expected 2 classes, but found {len(class_names)}: {class_names}"
    )

print("Class mapping:")
for i, name in enumerate(class_names):
    print(f"{i} = {name}")

# 6. MATCH IMAGE FILES WITH YOLO LABELS
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def find_image_label_pairs(split):
    image_dir = os.path.join(DATASET_PATH, split, "images")
    label_dir = os.path.join(DATASET_PATH, split, "labels")

    if not os.path.isdir(image_dir):
        raise FileNotFoundError(f"Image folder missing: {image_dir}")

    if not os.path.isdir(label_dir):
        raise FileNotFoundError(f"Label folder missing: {label_dir}")

    image_files = []
    for ext in IMAGE_EXTENSIONS:
        image_files.extend(
            glob.glob(os.path.join(image_dir, "**", f"*{ext}"),
                      recursive=True)
        )

    label_files = glob.glob(
        os.path.join(label_dir, "**", "*.txt"),
        recursive=True
    )

    # Match by relative filename stem.
    label_lookup = {}
    for label_path in label_files:
        rel = os.path.relpath(label_path, label_dir)
        stem = os.path.splitext(rel)[0]
        label_lookup[stem] = label_path

    pairs = []
    missing = []

    for image_path in image_files:
        rel = os.path.relpath(image_path, image_dir)
        stem = os.path.splitext(rel)[0]

        label_path = label_lookup.get(stem)

        if label_path is None:
            missing.append(image_path)
            continue

        with open(label_path, "r") as f:
            lines = [
                line.strip() for line in f.readlines()
                if line.strip()
            ]

        class_ids = []

        for line in lines:
            parts = line.split()
            if not parts:
                continue

            try:
                class_id = int(float(parts[0]))
                if 0 <= class_id < NUM_CLASSES:
                    class_ids.append(class_id)
            except ValueError:
                continue

        if not class_ids:
            continue

        # Image-level classification label:
        # majority class among YOLO objects in the image.
        counts = np.bincount(class_ids, minlength=NUM_CLASSES)
        image_label = int(np.argmax(counts))

        pairs.append((image_path, image_label))

    print(f"\n{split.upper()} split:")
    print("Images found:", len(image_files))
    print("Matched labeled images:", len(pairs))
    print("Images without matching label:", len(missing))

    if missing:
        print("First few unmatched images:")
        for p in missing[:5]:
            print(" -", os.path.basename(p))

    if not pairs:
        raise ValueError(
            f"No labeled images found in {split}. "
            "Check your image/label folder structure and filenames."
        )

    return pairs

train_pairs = find_image_label_pairs("train")
valid_pairs = find_image_label_pairs("valid")
test_pairs = find_image_label_pairs("test")

# 7. CHECK CLASS DISTRIBUTION
def show_distribution(pairs, split):
    labels = [label for _, label in pairs]
    counts = np.bincount(labels, minlength=NUM_CLASSES)

    print(f"\n{split} class distribution:")
    for i, name in enumerate(class_names):
        print(f"{name}: {counts[i]}")

show_distribution(train_pairs, "Training")
show_distribution(valid_pairs, "Validation")
show_distribution(test_pairs, "Testing")

# 8. CREATE TENSORFLOW DATASETS
def load_image(image_path, label):
    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(image, tf.float32)
    image = preprocess_input(image)

    label = tf.cast(label, tf.float32)

    return image, label

def create_dataset(pairs, training=False):
    paths = [p[0] for p in pairs]
    labels = [p[1] for p in pairs]

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=len(pairs),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset

train_ds = create_dataset(train_pairs, training=True)
valid_ds = create_dataset(valid_pairs)
test_ds = create_dataset(test_pairs)

# 9. BUILD DENSENET121 MODEL
print("\nBuilding DenseNet121 model...")

base_model = DenseNet121(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Freeze the complete pretrained backbone.
base_model.trainable = False

inputs = Input(shape=(IMG_SIZE, IMG_SIZE, 3))

x = base_model(inputs, training=False)
x = GlobalAveragePooling2D()(x)
x = BatchNormalization()(x)
x = Dense(256, activation="relu")(x)
x = Dropout(0.4)(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.3)(x)

outputs = Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 10. CALLBACKS
checkpoint_path = os.path.join(
    OUTPUT_PATH,
    "DenseNet121_Best.keras"
)

callbacks = [
    ModelCheckpoint(
        checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-7,
        verbose=1
    ),

    CSVLogger(
        os.path.join(OUTPUT_PATH, "training_log.csv")
    )
]

# 11. TRAIN MODEL
print("\nStarting DenseNet121 training...")

history = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

# 12. LOAD BEST MODEL
print("\nLoading best saved model...")

model = tf.keras.models.load_model(checkpoint_path)

# 13. TEST EVALUATION
print("\nEvaluating on test dataset...")

test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

y_true = np.array([label for _, label in test_pairs])

y_prob = model.predict(test_ds, verbose=1)
y_pred = np.argmax(y_prob, axis=1)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(
    y_true, y_pred, average="weighted", zero_division=0
)
recall = recall_score(
    y_true, y_pred, average="weighted", zero_division=0
)
f1 = f1_score(
    y_true, y_pred, average="weighted", zero_division=0
)

try:
    auc = roc_auc_score(
        y_true,
        y_prob[:, 1]
    )
except Exception:
    auc = float("nan")

print("\n========== TEST RESULTS ==========")
print(f"Test Loss     : {test_loss:.4f}")
print(f"Accuracy      : {accuracy * 100:.2f}%")
print(f"Precision     : {precision * 100:.2f}%")
print(f"Recall        : {recall * 100:.2f}%")
print(f"F1 Score      : {f1 * 100:.2f}%")
print(
    f"ROC-AUC       : {auc * 100:.2f}%"
    if not np.isnan(auc)
    else "ROC-AUC       : Not available"
)

# 14. CLASSIFICATION REPORT
report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=class_names,
    zero_division=0
)

print("\nClassification Report:")
print(report)

with open(
    os.path.join(OUTPUT_PATH, "classification_report.txt"),
    "w"
) as f:
    f.write(report)

# 15. CONFUSION MATRIX
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

fig, ax = plt.subplots(figsize=(7, 6))
disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)
plt.title("DenseNet121 - Confusion Matrix")
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "confusion_matrix.png"),
    dpi=300
)
plt.show()

# 16. ACCURACY GRAPH
plt.figure(figsize=(9, 5))
plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)
plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)
plt.title("DenseNet121 - Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "accuracy_graph.png"),
    dpi=300
)
plt.show()

# 17. LOSS GRAPH
plt.figure(figsize=(9, 5))
plt.plot(
    history.history["loss"],
    label="Training Loss"
)
plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)
plt.title("DenseNet121 - Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "loss_graph.png"),
    dpi=300
)
plt.show()

# 18. SAVE FINAL MODEL
final_model_path = os.path.join(
    OUTPUT_PATH,
    "DenseNet121_Final.keras"
)

model.save(final_model_path)

# 19. SAVE METRICS
metrics = {
    "Model": "DenseNet121",
    "Test Loss": float(test_loss),
    "Accuracy": float(accuracy * 100),
    "Precision": float(precision * 100),
    "Recall": float(recall * 100),
    "F1 Score": float(f1 * 100),
    "ROC-AUC": (
        float(auc * 100)
        if not np.isnan(auc)
        else None
    ),
    "Epochs": EPOCHS,
    "Batch Size": BATCH_SIZE,
    "Image Size": IMG_SIZE
}

metrics_df = pd.DataFrame([metrics])

metrics_df.to_csv(
    os.path.join(OUTPUT_PATH, "DenseNet121_metrics.csv"),
    index=False
)

with open(
    os.path.join(OUTPUT_PATH, "DenseNet121_metrics.json"),
    "w"
) as f:
    json.dump(metrics, f, indent=4)

# 20. SAVE TRAINING HISTORY
pd.DataFrame(history.history).to_csv(
    os.path.join(OUTPUT_PATH, "DenseNet121_history.csv"),
    index=False
)

# 21. SINGLE IMAGE PREDICTION FUNCTION
def predict_drowsiness(image_path):
    image = tf.io.read_file(image_path)
    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(image, tf.float32)
    image = preprocess_input(image)

    image = tf.expand_dims(image, axis=0)

    probabilities = model.predict(image, verbose=0)[0]

    predicted_class = int(np.argmax(probabilities))
    confidence = float(probabilities[predicted_class]) * 100

    print("\nPrediction:", class_names[predicted_class])
    print(f"Confidence: {confidence:.2f}%")

    for i, name in enumerate(class_names):
        print(f"{name}: {probabilities[i] * 100:.2f}%")

    return class_names[predicted_class], confidence

# 22. FINAL SUMMARY
print("\n" + "=" * 55)
print("DENSENET121 TRAINING COMPLETED")
print("=" * 55)

print(f"Test Accuracy : {accuracy * 100:.2f}%")
print(f"Test Precision: {precision * 100:.2f}%")
print(f"Test Recall   : {recall * 100:.2f}%")
print(f"Test F1 Score : {f1 * 100:.2f}%")

print("\nAll results saved at:")
print(OUTPUT_PATH)

print("\nSaved files:")
for filename in sorted(os.listdir(OUTPUT_PATH)):
    print(" -", filename)
```

```python

# ==============================================================
# DRIVER DROWSINESS DETECTION USING IMPROVED EFFICIENTNETB0
# Stage 1: Transfer Learning (30 epochs)
# Stage 2: Fine Tuning (15 epochs)
# Class Weights + Data Augmentation
# ==============================================================

# 1. IMPORT LIBRARIES
import os
import glob
import json
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import yaml

from google.colab import drive

from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.layers import (
    Input, Dense, Dropout, GlobalAveragePooling2D,
    BatchNormalization, RandomFlip, RandomRotation,
    RandomZoom, RandomContrast
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import (
    ModelCheckpoint, ReduceLROnPlateau, CSVLogger
)

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

# 2. REPRODUCIBILITY
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# 3. MOUNT GOOGLE DRIVE
drive.mount('/content/drive')

# 4. CONFIGURATION
DATASET_PATH = "/content/drive/MyDrive/Driver_test/Driver_testing"

OUTPUT_PATH = "/content/drive/MyDrive/Driver_test/EfficientNetB0_Improved_Results"

IMG_SIZE = 224
BATCH_SIZE = 16

INITIAL_EPOCHS = 30
FINE_TUNE_EPOCHS = 15

NUM_CLASSES = 2

os.makedirs(OUTPUT_PATH, exist_ok=True)

print("TensorFlow:", tf.__version__)
print("Dataset:", DATASET_PATH)
print("Results:", OUTPUT_PATH)

# 5. READ CLASS NAMES FROM DATA.YAML
yaml_path = os.path.join(DATASET_PATH, "data.yaml")

if not os.path.exists(yaml_path):
    raise FileNotFoundError(
        "data.yaml not found. Check your dataset folder."
    )

with open(yaml_path, "r") as f:
    yaml_data = yaml.safe_load(f)

class_names = yaml_data.get("names", ["Awake", "Drowsy"])

if isinstance(class_names, dict):
    class_names = [
        class_names[k] for k in sorted(class_names.keys())
    ]

if len(class_names) != NUM_CLASSES:
    raise ValueError(
        f"Expected 2 classes, found {len(class_names)}: {class_names}"
    )

print("\nClass mapping:")
for i, name in enumerate(class_names):
    print(f"{i}: {name}")

# 6. LOAD YOLO IMAGE-LABEL PAIRS
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def find_image_label_pairs(split):

    image_dir = os.path.join(DATASET_PATH, split, "images")
    label_dir = os.path.join(DATASET_PATH, split, "labels")

    if not os.path.isdir(image_dir):
        raise FileNotFoundError(image_dir)

    if not os.path.isdir(label_dir):
        raise FileNotFoundError(label_dir)

    image_files = []

    for ext in IMAGE_EXTENSIONS:
        image_files.extend(
            glob.glob(
                os.path.join(image_dir, "**", f"*{ext}"),
                recursive=True
            )
        )

    label_files = glob.glob(
        os.path.join(label_dir, "**", "*.txt"),
        recursive=True
    )

    label_lookup = {}

    for label_path in label_files:
        rel = os.path.relpath(label_path, label_dir)
        stem = os.path.splitext(rel)[0]
        label_lookup[stem] = label_path

    pairs = []
    missing = []

    for image_path in image_files:

        rel = os.path.relpath(image_path, image_dir)
        stem = os.path.splitext(rel)[0]

        label_path = label_lookup.get(stem)

        if label_path is None:
            missing.append(image_path)
            continue

        with open(label_path, "r") as f:
            lines = [
                line.strip()
                for line in f.readlines()
                if line.strip()
            ]

        class_ids = []

        for line in lines:
            parts = line.split()

            if not parts:
                continue

            try:
                class_id = int(float(parts[0]))

                if 0 <= class_id < NUM_CLASSES:
                    class_ids.append(class_id)

            except ValueError:
                continue

        if not class_ids:
            continue

        # Image-level label based on majority class.
        counts = np.bincount(
            class_ids,
            minlength=NUM_CLASSES
        )

        image_label = int(np.argmax(counts))

        pairs.append((image_path, image_label))

    print(f"\n{split.upper()} DATA")
    print("Total images:", len(image_files))
    print("Matched images:", len(pairs))
    print("Missing annotation files:", len(missing))

    if not pairs:
        raise ValueError(
            f"No valid labeled images found in {split}."
        )

    return pairs


train_pairs = find_image_label_pairs("train")
valid_pairs = find_image_label_pairs("valid")
test_pairs = find_image_label_pairs("test")

# 7. CHECK CLASS DISTRIBUTION
def show_distribution(pairs, split):

    labels = np.array([label for _, label in pairs])

    counts = np.bincount(
        labels,
        minlength=NUM_CLASSES
    )

    print(f"\n{split} distribution:")

    for i, name in enumerate(class_names):
        print(f"{name}: {counts[i]}")

show_distribution(train_pairs, "Training")
show_distribution(valid_pairs, "Validation")
show_distribution(test_pairs, "Testing")

# 8. CLASS WEIGHTS
train_labels = np.array(
    [label for _, label in train_pairs]
)

present_classes = np.unique(train_labels)

weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=train_labels
)

class_weights = {
    int(cls): float(weight)
    for cls, weight in zip(present_classes, weights)
}

print("\nClass weights:", class_weights)

# 9. IMAGE LOADING
def load_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    # EfficientNetB0 includes its own rescaling layer.
    # Keep image values in the 0-255 range.
    image = tf.cast(image, tf.float32)

    label = tf.cast(label, tf.int32)

    return image, label


def create_dataset(pairs, training=False):

    paths = [p[0] for p in pairs]
    labels = [p[1] for p in pairs]

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=len(pairs),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset


train_ds = create_dataset(train_pairs, training=True)
valid_ds = create_dataset(valid_pairs)
test_ds = create_dataset(test_pairs)

# 10. DATA AUGMENTATION
data_augmentation = tf.keras.Sequential(
    [
        RandomFlip("horizontal"),
        RandomRotation(0.05),
        RandomZoom(0.10),
        RandomContrast(0.10)
    ],
    name="data_augmentation"
)

# 11. BUILD EFFICIENTNETB0
print("\nBuilding EfficientNetB0...")

base_model = EfficientNetB0(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Initially freeze the entire backbone.
base_model.trainable = False

inputs = Input(
    shape=(IMG_SIZE, IMG_SIZE, 3)
)

x = data_augmentation(inputs)

x = base_model(
    x,
    training=False
)

x = GlobalAveragePooling2D()(x)
x = BatchNormalization()(x)

x = Dense(
    256,
    activation="relu"
)(x)

x = Dropout(0.4)(x)

x = Dense(
    128,
    activation="relu"
)(x)

x = Dropout(0.3)(x)

outputs = Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

# 12. STAGE 1 COMPILATION
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 13. STAGE 1 CALLBACKS
stage1_path = os.path.join(
    OUTPUT_PATH,
    "EfficientNetB0_Stage1_Best.keras"
)

stage1_callbacks = [

    ModelCheckpoint(
        stage1_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-7,
        verbose=1
    ),

    CSVLogger(
        os.path.join(OUTPUT_PATH, "stage1_training_log.csv")
    )
]

# 14. STAGE 1 TRAINING
print("\n" + "=" * 55)
print("STAGE 1: TRAINING CLASSIFICATION HEAD")
print("=" * 55)

history1 = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=INITIAL_EPOCHS,
    class_weight=class_weights,
    callbacks=stage1_callbacks,
    verbose=1
)

# 15. LOAD BEST STAGE 1 MODEL
print("\nLoading best Stage 1 model...")

model = tf.keras.models.load_model(stage1_path)

# 16. STAGE 2: FINE-TUNING
print("\n" + "=" * 55)
print("STAGE 2: FINE-TUNING TOP EFFICIENTNETB0 LAYERS")
print("=" * 55)

base_model = None

for layer in model.layers:
    if isinstance(layer, tf.keras.Model):
        if "efficientnet" in layer.name.lower():
            base_model = layer
            break

if base_model is None:
    raise RuntimeError(
        "Could not locate EfficientNetB0 backbone."
    )

base_model.trainable = True

# Freeze all layers except the last 30 non-BatchNorm layers.
for layer in base_model.layers[:-30]:
    layer.trainable = False

for layer in base_model.layers[-30:]:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False
    else:
        layer.trainable = True

print(
    "Trainable backbone layers:",
    sum(layer.trainable for layer in base_model.layers)
)

# Recompile with a very low learning rate.
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

stage2_path = os.path.join(
    OUTPUT_PATH,
    "EfficientNetB0_Stage2_Best.keras"
)

stage2_callbacks = [

    ModelCheckpoint(
        stage2_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-8,
        verbose=1
    ),

    CSVLogger(
        os.path.join(OUTPUT_PATH, "stage2_training_log.csv")
    )
]

# 17. STAGE 2 TRAINING
history2 = model.fit(
    train_ds,
    validation_data=valid_ds,
    initial_epoch=INITIAL_EPOCHS,
    epochs=INITIAL_EPOCHS + FINE_TUNE_EPOCHS,
    class_weight=class_weights,
    callbacks=stage2_callbacks,
    verbose=1
)

# 18. SELECT BEST MODEL USING VALIDATION LOSS
print("\nComparing Stage 1 and Stage 2 models...")

stage1_model = tf.keras.models.load_model(stage1_path)
stage2_model = tf.keras.models.load_model(stage2_path)

stage1_loss = stage1_model.evaluate(
    valid_ds,
    verbose=0
)[0]

stage2_loss = stage2_model.evaluate(
    valid_ds,
    verbose=0
)[0]

print(f"Stage 1 validation loss: {stage1_loss:.4f}")
print(f"Stage 2 validation loss: {stage2_loss:.4f}")

if stage2_loss < stage1_loss:
    model = stage2_model
    print("Selected Stage 2 model.")
else:
    model = stage1_model
    print("Selected Stage 1 model.")

# 19. SAVE SELECTED MODEL
final_model_path = os.path.join(
    OUTPUT_PATH,
    "EfficientNetB0_Final.keras"
)

model.save(final_model_path)

# 20. FINAL TEST EVALUATION
print("\n" + "=" * 55)
print("FINAL TEST EVALUATION")
print("=" * 55)

test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

y_true = np.array(
    [label for _, label in test_pairs]
)

y_prob = model.predict(
    test_ds,
    verbose=1
)

y_pred = np.argmax(y_prob, axis=1)

accuracy = accuracy_score(y_true, y_pred)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

try:
    auc = roc_auc_score(
        y_true,
        y_prob[:, 1]
    )
except Exception:
    auc = float("nan")

print("\n========== FINAL RESULTS ==========")
print(f"Test Loss : {test_loss:.4f}")
print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")

if not np.isnan(auc):
    print(f"ROC-AUC   : {auc * 100:.2f}%")
else:
    print("ROC-AUC   : Not available")

# 21. CLASSIFICATION REPORT
report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=class_names,
    zero_division=0
)

print("\nClassification Report:")
print(report)

with open(
    os.path.join(OUTPUT_PATH, "classification_report.txt"),
    "w"
) as f:
    f.write(report)

# 22. CONFUSION MATRIX
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

fig, ax = plt.subplots(figsize=(7, 6))

disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)

plt.title("EfficientNetB0 - Confusion Matrix")
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "confusion_matrix.png"),
    dpi=300
)

plt.show()

# 23. COMBINE TRAINING HISTORIES
combined_history = {}

for key in history1.history.keys():
    combined_history[key] = (
        history1.history[key] +
        history2.history.get(key, [])
    )

# 24. ACCURACY GRAPH
plt.figure(figsize=(10, 5))

plt.plot(
    combined_history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    combined_history["val_accuracy"],
    label="Validation Accuracy"
)

plt.axvline(
    x=INITIAL_EPOCHS - 1,
    linestyle="--",
    label="Fine-Tuning Starts"
)

plt.title("EfficientNetB0 - Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "accuracy_graph.png"),
    dpi=300
)

plt.show()

# 25. LOSS GRAPH
plt.figure(figsize=(10, 5))

plt.plot(
    combined_history["loss"],
    label="Training Loss"
)

plt.plot(
    combined_history["val_loss"],
    label="Validation Loss"
)

plt.axvline(
    x=INITIAL_EPOCHS - 1,
    linestyle="--",
    label="Fine-Tuning Starts"
)

plt.title("EfficientNetB0 - Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "loss_graph.png"),
    dpi=300
)

plt.show()

# 26. SAVE TRAINING HISTORY
pd.DataFrame(combined_history).to_csv(
    os.path.join(OUTPUT_PATH, "training_history.csv"),
    index=False
)

# 27. SAVE METRICS
metrics = {
    "Model": "EfficientNetB0 Improved",
    "Test Loss": float(test_loss),
    "Accuracy": float(accuracy * 100),
    "Precision": float(precision * 100),
    "Recall": float(recall * 100),
    "F1 Score": float(f1 * 100),
    "ROC-AUC": (
        float(auc * 100)
        if not np.isnan(auc)
        else None
    ),
    "Initial Epochs": INITIAL_EPOCHS,
    "Fine Tuning Epochs": FINE_TUNE_EPOCHS,
    "Batch Size": BATCH_SIZE,
    "Image Size": IMG_SIZE
}

pd.DataFrame([metrics]).to_csv(
    os.path.join(OUTPUT_PATH, "final_metrics.csv"),
    index=False
)

with open(
    os.path.join(OUTPUT_PATH, "final_metrics.json"),
    "w"
) as f:
    json.dump(metrics, f, indent=4)

# 28. SINGLE IMAGE PREDICTION
def predict_drowsiness(image_path):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(image, tf.float32)
    image = tf.expand_dims(image, axis=0)

    probabilities = model.predict(
        image,
        verbose=0
    )[0]

    predicted_class = int(np.argmax(probabilities))

    confidence = float(
        probabilities[predicted_class] * 100
    )

    print("\nPrediction:", class_names[predicted_class])
    print(f"Confidence: {confidence:.2f}%")

    for i, name in enumerate(class_names):
        print(
            f"{name}: {probabilities[i] * 100:.2f}%"
        )

    return class_names[predicted_class], confidence

# 29. FINAL SUMMARY
print("\n" + "=" * 60)
print("EFFICIENTNETB0 IMPROVED TRAINING COMPLETED")
print("=" * 60)

print(f"Final Test Accuracy : {accuracy * 100:.2f}%")
print(f"Precision           : {precision * 100:.2f}%")
print(f"Recall              : {recall * 100:.2f}%")
print(f"F1 Score            : {f1 * 100:.2f}%")

print("\nResults saved at:")
print(OUTPUT_PATH)

print("\nSaved files:")

for filename in sorted(os.listdir(OUTPUT_PATH)):
    print(" -", filename)
```

```python

# ================================================================
# DRIVER DROWSINESS DETECTION
# HYBRID ENSEMBLE: EFFICIENTNETB0 + RESNET50
# Google Colab - Complete Single Cell Code
# Stage 1: 25 epochs
# Stage 2: 10 epochs fine-tuning
# ================================================================

import os
import glob
import json
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import yaml

from google.colab import drive

from tensorflow.keras.applications import EfficientNetB0, ResNet50
from tensorflow.keras.layers import (
    Input, Dense, Dropout, BatchNormalization,
    GlobalAveragePooling2D, Concatenate,
    RandomFlip, RandomRotation, RandomZoom,
    RandomContrast, Lambda
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import (
    ModelCheckpoint, ReduceLROnPlateau, CSVLogger
)

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report,
    confusion_matrix, ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

# 1. REPRODUCIBILITY
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# 2. MOUNT GOOGLE DRIVE
drive.mount('/content/drive')

# 3. CONFIGURATION
DATASET_PATH = "/content/drive/MyDrive/Driver_test/Driver_testing"

OUTPUT_PATH = "/content/drive/MyDrive/Driver_test/EfficientNetB0_ResNet50_Ensemble_Results"

IMG_SIZE = 224
BATCH_SIZE = 8
INITIAL_EPOCHS = 25
FINE_TUNE_EPOCHS = 10
NUM_CLASSES = 2

os.makedirs(OUTPUT_PATH, exist_ok=True)

print("TensorFlow version:", tf.__version__)
print("Dataset:", DATASET_PATH)
print("Results:", OUTPUT_PATH)

# 4. READ DATA.YAML
yaml_path = os.path.join(DATASET_PATH, "data.yaml")

if not os.path.exists(yaml_path):
    raise FileNotFoundError("data.yaml not found!")

with open(yaml_path, "r") as f:
    yaml_data = yaml.safe_load(f)

class_names = yaml_data.get("names", ["Awake", "Drowsy"])

if isinstance(class_names, dict):
    class_names = [
        class_names[k] for k in sorted(class_names.keys())
    ]

if len(class_names) != NUM_CLASSES:
    raise ValueError(
        f"Expected 2 classes, found {len(class_names)}: {class_names}"
    )

print("\nClass mapping:")
for i, name in enumerate(class_names):
    print(f"{i} = {name}")

# 5. MATCH IMAGES AND YOLO LABELS
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def find_image_label_pairs(split):

    image_dir = os.path.join(DATASET_PATH, split, "images")
    label_dir = os.path.join(DATASET_PATH, split, "labels")

    if not os.path.isdir(image_dir):
        raise FileNotFoundError(image_dir)

    if not os.path.isdir(label_dir):
        raise FileNotFoundError(label_dir)

    image_files = []

    for ext in IMAGE_EXTENSIONS:
        image_files.extend(
            glob.glob(
                os.path.join(image_dir, "**", f"*{ext}"),
                recursive=True
            )
        )

    label_files = glob.glob(
        os.path.join(label_dir, "**", "*.txt"),
        recursive=True
    )

    label_lookup = {}

    for label_path in label_files:
        rel = os.path.relpath(label_path, label_dir)
        stem = os.path.splitext(rel)[0]
        label_lookup[stem] = label_path

    pairs = []
    missing = []

    for image_path in image_files:

        rel = os.path.relpath(image_path, image_dir)
        stem = os.path.splitext(rel)[0]

        label_path = label_lookup.get(stem)

        if label_path is None:
            missing.append(image_path)
            continue

        with open(label_path, "r") as f:
            lines = [
                line.strip()
                for line in f.readlines()
                if line.strip()
            ]

        class_ids = []

        for line in lines:
            parts = line.split()

            if not parts:
                continue

            try:
                class_id = int(float(parts[0]))

                if 0 <= class_id < NUM_CLASSES:
                    class_ids.append(class_id)

            except ValueError:
                continue

        # Skip images with empty or invalid annotations.
        if not class_ids:
            continue

        # Image-level label from majority YOLO class.
        counts = np.bincount(
            class_ids,
            minlength=NUM_CLASSES
        )

        image_label = int(np.argmax(counts))

        pairs.append((image_path, image_label))

    print(f"\n{split.upper()} SPLIT")
    print("Images found:", len(image_files))
    print("Matched labeled images:", len(pairs))
    print("Missing annotation files:", len(missing))

    if not pairs:
        raise ValueError(f"No labeled images found in {split}.")

    return pairs


train_pairs = find_image_label_pairs("train")
valid_pairs = find_image_label_pairs("valid")
test_pairs = find_image_label_pairs("test")

# 6. CLASS DISTRIBUTION
def show_distribution(pairs, split):

    labels = np.array([label for _, label in pairs])

    counts = np.bincount(
        labels,
        minlength=NUM_CLASSES
    )

    print(f"\n{split} class distribution:")

    for i, name in enumerate(class_names):
        print(f"{name}: {counts[i]}")

show_distribution(train_pairs, "Training")
show_distribution(valid_pairs, "Validation")
show_distribution(test_pairs, "Testing")

# 7. CLASS WEIGHTS
train_labels = np.array(
    [label for _, label in train_pairs]
)

present_classes = np.unique(train_labels)

weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=train_labels
)

class_weights = {
    int(c): float(w)
    for c, w in zip(present_classes, weights)
}

print("\nClass weights:", class_weights)

# 8. IMAGE LOADING
def load_image(image_path, label):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(image, tf.float32)
    label = tf.cast(label, tf.int32)

    # Keep image pixels in 0-255 range.
    # Each backbone gets its own preprocessing.

    return image, label


def create_dataset(pairs, training=False):

    paths = [p[0] for p in pairs]
    labels = [p[1] for p in pairs]

    ds = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:
        ds = ds.shuffle(
            len(pairs),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    ds = ds.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(tf.data.AUTOTUNE)

    return ds


train_ds = create_dataset(train_pairs, training=True)
valid_ds = create_dataset(valid_pairs)
test_ds = create_dataset(test_pairs)

# 9. DATA AUGMENTATION
augmentation = tf.keras.Sequential(
    [
        RandomFlip("horizontal"),
        RandomRotation(0.04),
        RandomZoom(0.08),
        RandomContrast(0.08)
    ],
    name="data_augmentation"
)

# 10. BUILD BOTH PRETRAINED BACKBONES
print("\nLoading EfficientNetB0 and ResNet50...")

efficientnet = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

resnet = ResNet50(
    include_top=False,
    weights="imagenet",
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

efficientnet.trainable = False
resnet.trainable = False

# 11. HYBRID ENSEMBLE MODEL
inputs = Input(
    shape=(IMG_SIZE, IMG_SIZE, 3),
    name="input_image"
)

augmented = augmentation(inputs)

# EfficientNetB0 expects RGB pixels in 0-255.
efficient_features = efficientnet(
    augmented,
    training=False
)

efficient_features = GlobalAveragePooling2D(
    name="efficientnet_pool"
)(efficient_features)

# ResNet50 requires RGB-to-BGR and ImageNet mean subtraction.
resnet_input = Lambda(
    tf.keras.applications.resnet50.preprocess_input,
    name="resnet_preprocessing"
)(augmented)

resnet_features = resnet(
    resnet_input,
    training=False
)

resnet_features = GlobalAveragePooling2D(
    name="resnet_pool"
)(resnet_features)

# Feature-level fusion
combined_features = Concatenate(
    name="feature_fusion"
)([efficient_features, resnet_features])

x = BatchNormalization()(combined_features)

x = Dense(
    512,
    activation="relu"
)(x)

x = Dropout(0.5)(x)

x = Dense(
    256,
    activation="relu"
)(x)

x = BatchNormalization()(x)
x = Dropout(0.35)(x)

outputs = Dense(
    NUM_CLASSES,
    activation="softmax",
    name="classification"
)(x)

model = Model(
    inputs=inputs,
    outputs=outputs,
    name="EfficientNetB0_ResNet50_Ensemble"
)

# 12. STAGE 1 COMPILATION
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 13. STAGE 1 CALLBACKS
stage1_path = os.path.join(
    OUTPUT_PATH,
    "Ensemble_Stage1_Best.keras"
)

callbacks1 = [
    ModelCheckpoint(
        stage1_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-7,
        verbose=1
    ),

    CSVLogger(
        os.path.join(OUTPUT_PATH, "stage1_log.csv")
    )
]

# 14. STAGE 1 TRAINING
print("\n" + "=" * 60)
print("STAGE 1: TRAINING ENSEMBLE CLASSIFICATION HEAD")
print("=" * 60)

history1 = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=INITIAL_EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks1,
    verbose=1
)

# 15. LOAD BEST STAGE 1 MODEL
print("\nLoading best Stage 1 checkpoint...")

model = tf.keras.models.load_model(
    stage1_path,
    custom_objects={
        "preprocess_input":
        tf.keras.applications.resnet50.preprocess_input
    }
)

# 16. FINE-TUNE BOTH BACKBONES
print("\n" + "=" * 60)
print("STAGE 2: FINE-TUNING BOTH BACKBONES")
print("=" * 60)

efficientnet = model.get_layer("efficientnetb0")
resnet = model.get_layer("resnet50")

efficientnet.trainable = True
resnet.trainable = True

# Unfreeze only the final 30 EfficientNet layers.
for layer in efficientnet.layers[:-30]:
    layer.trainable = False

for layer in efficientnet.layers[-30:]:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False
    else:
        layer.trainable = True

# Unfreeze only the final 30 ResNet layers.
for layer in resnet.layers[:-30]:
    layer.trainable = False

for layer in resnet.layers[-30:]:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False
    else:
        layer.trainable = True

print(
    "EfficientNet trainable layers:",
    sum(l.trainable for l in efficientnet.layers)
)

print(
    "ResNet trainable layers:",
    sum(l.trainable for l in resnet.layers)
)

# 17. RECOMPILE WITH LOW LEARNING RATE
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 18. STAGE 2 CALLBACKS
stage2_path = os.path.join(
    OUTPUT_PATH,
    "Ensemble_Stage2_Best.keras"
)

callbacks2 = [
    ModelCheckpoint(
        stage2_path,
        monitor="val_loss",
        save_best_only=True,
        mode="min",
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-8,
        verbose=1
    ),

    CSVLogger(
        os.path.join(OUTPUT_PATH, "stage2_log.csv")
    )
]

# 19. STAGE 2 TRAINING
history2 = model.fit(
    train_ds,
    validation_data=valid_ds,
    initial_epoch=INITIAL_EPOCHS,
    epochs=INITIAL_EPOCHS + FINE_TUNE_EPOCHS,
    class_weight=class_weights,
    callbacks=callbacks2,
    verbose=1
)

# 20. SELECT BEST CHECKPOINT BY VALIDATION LOSS
print("\nSelecting best model...")

stage1_model = tf.keras.models.load_model(
    stage1_path,
    custom_objects={
        "preprocess_input":
        tf.keras.applications.resnet50.preprocess_input
    }
)

stage2_model = tf.keras.models.load_model(
    stage2_path,
    custom_objects={
        "preprocess_input":
        tf.keras.applications.resnet50.preprocess_input
    }
)

stage1_val_loss = stage1_model.evaluate(
    valid_ds,
    verbose=0
)[0]

stage2_val_loss = stage2_model.evaluate(
    valid_ds,
    verbose=0
)[0]

print("Stage 1 validation loss:", stage1_val_loss)
print("Stage 2 validation loss:", stage2_val_loss)

if stage2_val_loss < stage1_val_loss:
    model = stage2_model
    print("Selected Stage 2 ensemble.")
else:
    model = stage1_model
    print("Selected Stage 1 ensemble.")

# 21. SAVE FINAL MODEL
final_model_path = os.path.join(
    OUTPUT_PATH,
    "EfficientNetB0_ResNet50_Ensemble_Final.keras"
)

model.save(final_model_path)

# 22. TEST EVALUATION
print("\n" + "=" * 60)
print("FINAL TEST EVALUATION")
print("=" * 60)

test_loss, test_acc = model.evaluate(
    test_ds,
    verbose=1
)

y_true = np.array(
    [label for _, label in test_pairs]
)

y_prob = model.predict(
    test_ds,
    verbose=1
)

y_pred = np.argmax(y_prob, axis=1)

accuracy = accuracy_score(y_true, y_pred)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

try:
    auc = roc_auc_score(
        y_true,
        y_prob[:, 1]
    )
except Exception:
    auc = float("nan")

print("\n========== ENSEMBLE RESULTS ==========")
print(f"Test Loss : {test_loss:.4f}")
print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")

if not np.isnan(auc):
    print(f"ROC-AUC   : {auc * 100:.2f}%")
else:
    print("ROC-AUC   : Not available")

# 23. CLASSIFICATION REPORT
report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=class_names,
    zero_division=0
)

print("\nClassification Report:")
print(report)

with open(
    os.path.join(OUTPUT_PATH, "classification_report.txt"),
    "w"
) as f:
    f.write(report)

# 24. CONFUSION MATRIX
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

fig, ax = plt.subplots(figsize=(7, 6))

disp.plot(
    cmap="Blues",
    values_format="d",
    ax=ax
)

plt.title("EfficientNetB0 + ResNet50 Ensemble")
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "confusion_matrix.png"),
    dpi=300
)

plt.show()

# 25. COMBINE TRAINING HISTORY
combined_history = {}

for key in history1.history.keys():
    combined_history[key] = (
        history1.history[key] +
        history2.history.get(key, [])
    )

# 26. ACCURACY GRAPH
plt.figure(figsize=(10, 5))

plt.plot(
    combined_history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    combined_history["val_accuracy"],
    label="Validation Accuracy"
)

plt.axvline(
    x=INITIAL_EPOCHS - 1,
    linestyle="--",
    label="Fine-Tuning Starts"
)

plt.title("EfficientNetB0 + ResNet50 - Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "accuracy_graph.png"),
    dpi=300
)

plt.show()

# 27. LOSS GRAPH
plt.figure(figsize=(10, 5))

plt.plot(
    combined_history["loss"],
    label="Training Loss"
)

plt.plot(
    combined_history["val_loss"],
    label="Validation Loss"
)

plt.axvline(
    x=INITIAL_EPOCHS - 1,
    linestyle="--",
    label="Fine-Tuning Starts"
)

plt.title("EfficientNetB0 + ResNet50 - Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_PATH, "loss_graph.png"),
    dpi=300
)

plt.show()

# 28. SAVE HISTORY AND METRICS
pd.DataFrame(combined_history).to_csv(
    os.path.join(OUTPUT_PATH, "training_history.csv"),
    index=False
)

metrics = {
    "Model": "EfficientNetB0 + ResNet50 Ensemble",
    "Test Loss": float(test_loss),
    "Accuracy": float(accuracy * 100),
    "Precision": float(precision * 100),
    "Recall": float(recall * 100),
    "F1 Score": float(f1 * 100),
    "ROC-AUC": (
        float(auc * 100)
        if not np.isnan(auc)
        else None
    ),
    "Initial Epochs": INITIAL_EPOCHS,
    "Fine Tuning Epochs": FINE_TUNE_EPOCHS,
    "Batch Size": BATCH_SIZE,
    "Image Size": IMG_SIZE
}

pd.DataFrame([metrics]).to_csv(
    os.path.join(OUTPUT_PATH, "ensemble_metrics.csv"),
    index=False
)

with open(
    os.path.join(OUTPUT_PATH, "ensemble_metrics.json"),
    "w"
) as f:
    json.dump(metrics, f, indent=4)

# 29. SINGLE IMAGE PREDICTION
def predict_drowsiness(image_path):

    image = tf.io.read_file(image_path)

    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        [IMG_SIZE, IMG_SIZE]
    )

    image = tf.cast(image, tf.float32)
    image = tf.expand_dims(image, axis=0)

    probabilities = model.predict(
        image,
        verbose=0
    )[0]

    predicted_class = int(np.argmax(probabilities))

    confidence = float(
        probabilities[predicted_class] * 100
    )

    print("\nPrediction:", class_names[predicted_class])
    print(f"Confidence: {confidence:.2f}%")

    for i, name in enumerate(class_names):
        print(
            f"{name}: {probabilities[i] * 100:.2f}%"
        )

    return class_names[predicted_class], confidence

# 30. FINAL SUMMARY
print("\n" + "=" * 60)
print("ENSEMBLE TRAINING COMPLETED")
print("=" * 60)

print(f"Test Accuracy : {accuracy * 100:.2f}%")
print(f"Precision     : {precision * 100:.2f}%")
print(f"Recall        : {recall * 100:.2f}%")
print(f"F1 Score      : {f1 * 100:.2f}%")

print("\nResults saved in:")
print(OUTPUT_PATH)

print("\nSaved files:")

for filename in sorted(os.listdir(OUTPUT_PATH)):
    print(" -", filename)
```
